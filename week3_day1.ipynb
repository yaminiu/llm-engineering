```python
from IPython.display import display
from diffusers import AutoPipelineForText2Image
import torch

pipe = AutoPipelineForText2Image.from_pretrained("stabilityai/sdxl-turbo", dtype=torch.float16,variant="fp16")
pipe.to("cuda")
prompt = " A class of students learning AI Engineering in a virant pop-art style"
image = pipe(prompt=prompt, num_inference_steps=4, guidance_scale=0.0).image[0]
display(image)
```
